## 2. Your first program: `print()`
`print()` is a **function** that shows something on the screen. Whatever you put inside the round brackets is printed.
Text must be inside quotes - double `" "` or single `' '` - so that Python knows it is text and not a command.

Run the cell below with **Shift + Enter**.

````python
# print() shows the text inside the brackets on the screen
print("Hello, Bannari Amman!")      # text in double quotes
print('Welcome to Python')          # single quotes work the same way
print("I am learning Python for AI")
````

> **What happened?** Each `print()` showed its text on a **new line**. The quotes themselves are not printed - they only mark where the text starts and ends.

`print()` can also show **numbers** and the **result of a calculation**. Numbers do not need quotes.
You can print **several values at once** by separating them with commas.

````python
print(25)                              # a number - no quotes needed
print(10 + 5)                          # Python calculates first, then prints 15
print("My age is", 18)                 # text and a number, separated by a comma
print("Total marks:", 85 + 90 + 78)    # text and a calculation
````

> **What happened?** `10 + 5` was **calculated** and only the answer `15` was printed. When values are separated by commas, `print()` puts a **space** between them automatically.

### `sep` and `end`
`print()` has two useful settings:
- `sep` (separator) decides what goes **between** values. Default: a space.
- `end` decides what goes at the **end** of the line. Default: a new line, which is why each `print()` starts on a new line.

````python
print("01", "08", "2026", sep="-")   # put "-" between the values instead of a space
print("Loading", end="...")          # end with "..." instead of going to a new line
print("done!")                       # so this continues on the same line
````

> **What happened?** The date was joined with `-` because of `sep="-"`. The word `done!` appeared on the **same line** as `Loading...` because `end="..."` stopped the first `print()` from moving to a new line.

## 3. Variables
A **variable** is a named box in the computer's memory that stores a value. We use the `=` sign to put a value into the box.

```
name = "Asha"      # create a box called name and put "Asha" in it
```
**Note:** in Python, `=` means **store** (assign), not *equal to*. Read `age = 18` as *"age gets 18"*.

````python
# Create three variables and store values in them
name = "Asha"      # a piece of text
age = 18           # a whole number
cgpa = 8.6         # a decimal number

# Use the variable names to print their values (no quotes around variable names!)
print(name)
print(age)
print(cgpa)
print(name, "is", age, "years old and has a CGPA of", cgpa)
````

> **What happened?** Python remembered the values and printed them when we used the variable names. Notice: `print(name)` printed **Asha**, but `print("name")` would print the word **name**, because quotes turn it into plain text.

A variable can be **changed** at any time. Storing a new value **replaces** the old one - the box can hold only one value at a time.

````python
attendance = 70                   # first value
print("Attendance:", attendance)

attendance = 76                   # new value replaces 70
print("Attendance now:", attendance)
````

> **What happened?** The first `print()` showed 70. After the second assignment, the box `attendance` holds 76 and the old value 70 is gone.

### Rules for variable names
| Allowed | Not allowed |
|---|---|
| `marks`, `total_marks`, `mark1`, `_count` | `1mark` (cannot start with a number) |
| letters, numbers, underscore `_` | `total marks` (no spaces), `total-marks` (no `-`) |
| | Python keywords such as `if`, `for`, `while`, `class` |

- Names are **case sensitive**: `Marks` and `marks` are two different variables.
- Use meaningful names: `hostel_fee` is much better than `x`.
- Never use `print`, `input` or `type` as variable names - it breaks those functions.

### Multiple assignment and swapping
Python lets you give values to several variables in **one line**. It also has a very easy way to **swap** two values.

````python
a, b, c = 10, 20, 30     # a gets 10, b gets 20, c gets 30 (matched by position)
print(a, b, c)

x = 5
y = 9
x, y = y, x              # swap: x gets y's value and y gets x's value at the same time
print("x =", x, " y =", y)
````

> **What happened?** Values on the right are given to the names on the left **in order**. In the swap, Python first reads the right side (`9, 5`) and then stores them, so `x` becomes 9 and `y` becomes 5. In many other languages you need a third temporary variable to do this.

## 4. Data types
Every value in Python has a **type**. The type decides what you can do with the value (for example, you can divide numbers but not names).

| Type | Meaning | Examples |
|---|---|---|
| `int` | integer - whole numbers | `18`, `-5`, `2026` |
| `float` | numbers with a decimal point | `8.6`, `3.14`, `-0.5` |
| `str` | string - text in quotes | `"Asha"`, `'CSE'`, `"123"` |
| `bool` | boolean - only two values | `True`, `False` |

The function `type()` tells you the type of any value.

````python
roll_no = 101          # whole number      -> int
cgpa = 8.6             # decimal number    -> float
dept = "AI and DS"     # text in quotes    -> str
is_hostel = True       # True or False     -> bool (no quotes, capital T)

print(type(roll_no))
print(type(cgpa))
print(type(dept))
print(type(is_hostel))
````

> **What happened?** Python shows each type as `<class '...'>`. You will often check types like this when your program gives an unexpected result.

**Careful:** a number inside quotes, like `"123"`, is a **string** (text), not a number. Watch what `+` does with text and with numbers.

````python
print(type("123"))   # in quotes -> str
print(type(123))     # no quotes -> int
print("10" + "5")    # + on two strings JOINS them   -> "105"
print(10 + 5)        # + on two numbers ADDS them    -> 15
````

> **What happened?** `"10" + "5"` gave `105` because `+` **joins** (concatenates) strings. `10 + 5` gave `15` because `+` **adds** numbers. The same symbol does different jobs depending on the type - so types matter!
> 
> There is also a special value `None`, which means *no value yet*, for example `result = None`.

## 5. Type conversion
Sometimes we must change a value from one type to another. Python gives us conversion functions:
- `int()` - convert to a whole number
- `float()` - convert to a decimal number
- `str()` - convert to text

````python
marks_text = "87"              # this is text
marks = int(marks_text)        # convert text -> int so we can do maths
print(marks + 3)               # 90

price = float("49.50")         # convert text -> float
print(price * 2)               # 99.0

roll = 101
message = "Roll number " + str(roll)   # convert int -> str so it can be joined with text
print(message)

print(int(7.9))      # float -> int simply CUTS OFF the decimal part -> 7
print(round(7.9))    # round() rounds to the NEAREST whole number    -> 8
````

> **What happened?** After conversion we could do maths with `"87"` and join `101` with text. Note the difference: `int(7.9)` gives **7** (it chops), but `round(7.9)` gives **8** (it rounds).
> 
> Not every conversion works: `int("hello")` gives a `ValueError`, because "hello" is not a number.

## 6. Taking input from the user
`input()` pauses the program, shows a message, and waits for the user to type something and press **Enter**. Whatever the user types is stored in a variable.

In Colab, a small text box appears under the cell - click it, type, and press Enter.

````python
name = input("Enter your name: ")   # show the message, wait, store what is typed in name
print("Hello,", name)
````

> **What happened?** The text you typed was stored in `name` and then printed.

**Very important:** `input()` **always** gives a **string**, even if you type a number. If you want to calculate with it, convert it using `int()` or `float()` - you can do it in the same line.

````python
age = int(input("Enter your age: "))    # input gives text; int() converts it to a number
print("Next year you will be", age + 1) # now we can add 1
````

> **What happened?** Python works from the **inside out**: first `input()` reads the text (for example `"18"`), then `int()` turns it into the number `18`, then it is stored in `age`. Without `int()`, `age + 1` would give a `TypeError`.

## 7. Arithmetic operators
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide (always gives a float) | `7 / 2` | `3.5` |
| `//` | floor division (only the whole part) | `7 // 2` | `3` |
| `%` | modulus (the remainder) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

Run the cell and compare each result with the table.

````python
a = 17
b = 5
print("a + b  =", a + b)    # 22
print("a - b  =", a - b)    # 12
print("a * b  =", a * b)    # 85
print("a / b  =", a / b)    # 3.4  (normal division, gives a decimal)
print("a // b =", a // b)   # 3    (5 goes into 17 three whole times)
print("a % b  =", a % b)    # 2    (remainder: 17 - 3*5 = 2)
print("a ** b =", a ** b)   # 1419857 (17 to the power 5)
````

> **What happened?** `/` gives the exact answer `3.4`. `//` gives only the **whole part** `3`, and `%` gives what is **left over** `2`. Think of 17 chocolates shared among 5 friends: each gets `17 // 5 = 3` and `17 % 5 = 2` are left.

### Order of operations
Python follows the same rule you learned in maths (**BODMAS**): brackets first, then `**`, then `* / // %`, then `+ -`.
When in doubt, **use brackets** - they make your meaning clear.

````python
print(2 + 3 * 4)       # multiplication first: 2 + 12 = 14
print((2 + 3) * 4)     # brackets first: 5 * 4 = 20

# Average of three marks - brackets are essential!
m1, m2, m3 = 80, 90, 70
print("Wrong average:", m1 + m2 + m3 / 3)     # only m3 is divided by 3
print("Right average:", (m1 + m2 + m3) / 3)   # the whole total is divided by 3
````

> **What happened?** Without brackets Python calculated `80 + 90 + (70 / 3)` = 193.33, which is wrong. With brackets it first added all marks (240) and then divided by 3 = 80.0. A very common beginner mistake!

### Real example: splitting the mess bill
Four friends share a bill of Rs. 1450. How much does each pay? And how would the bill be paid with Rs. 100 notes?

````python
total_bill = 1450      # rupees
friends = 4

share = total_bill / friends          # normal division -> 362.5
print("Each person pays Rs.", share)

notes = total_bill // 100             # how many full 100-rupee notes -> 14
change = total_bill % 100             # what is left over             -> 50
print(notes, "notes of Rs.100 and Rs.", change, "in change")
````

> **What happened?** `/` gave the exact share. `//` and `%` together split 1450 into 14 hundreds and 50 rupees - the same idea is used to convert minutes into hours, or seconds into minutes.

### `%` is very useful
Two tricks you will use again and again:
- `n % 2 == 0` is True when `n` is **even** (no remainder when divided by 2).
- `n % 10` gives the **last digit** of `n`.

````python
n = 2026
print("Last digit:", n % 10)        # 2026 divided by 10 leaves remainder 6
print("Is even?", n % 2 == 0)       # remainder 0 when divided by 2 -> True
````

> **What happened?** 2026 ÷ 10 = 202 remainder **6**, so the last digit is 6. 2026 ÷ 2 leaves remainder 0, so it is even and the comparison gives `True`.

## 8. Assignment shortcuts
We often update a variable using its own old value. Python has shortcuts:

| Shortcut | Same as |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |

````python
steps = 0
steps += 1000      # walked to the canteen:  steps = 0 + 1000
steps += 2500      # walked to the library:  steps = 1000 + 2500
print("Steps today:", steps)

balance = 500
balance -= 120     # spent Rs. 120 on lunch: balance = 500 - 120
print("UPI balance:", balance)
````

> **What happened?** Each `+=` added to the old value, so steps became 3500. `-=` reduced the balance to 380. These shortcuts are used a lot inside loops (Notebook 2).

## 9. Comparison operators
A comparison asks a **question** whose answer is always `True` or `False` (a `bool`).

| Operator | Question |
|---|---|
| `==` | Are they equal? (**two** `=` signs) |
| `!=` | Are they different? |
| `>` `<` | Greater than? Less than? |
| `>=` `<=` | Greater than or equal? Less than or equal? |

````python
attendance = 78
print(attendance >= 75)    # is 78 at least 75?   -> True
print(attendance == 100)   # is it exactly 100?   -> False
print(attendance != 0)     # is it different from 0? -> True
print("cat" == "Cat")      # text comparison is case sensitive -> False
````

> **What happened?** Each line printed the answer to a yes/no question. Remember: `=` **stores** a value, `==` **compares** two values. Mixing them up is the most common beginner error.

## 10. Logical operators: `and`, `or`, `not`
They combine True/False answers:
- `and` - True only if **both** sides are True
- `or` - True if **at least one** side is True
- `not` - reverses the answer (True becomes False, False becomes True)

Example: a student can write the exam only if attendance is at least 75 **and** internal marks are at least 40.

````python
attendance = 80
internal_marks = 42

eligible = attendance >= 75 and internal_marks >= 40   # both conditions must be True
print("Eligible for exam:", eligible)

is_weekend = False
is_holiday = True
print("No class today:", is_weekend or is_holiday)        # one True is enough -> True
print("Class today:", not (is_weekend or is_holiday))     # not reverses it      -> False
````

> **What happened?** Both conditions were True, so `eligible` is True. Change `internal_marks` to 30 and run again - it becomes False because `and` needs **both** sides to be True.

### AI connection
In Week 1 you saw a spam filter. A very simple **rule-based** spam check could combine clues with `and`:

````python
# Three clues about an email (True = the clue is present)
has_link = True
from_unknown_sender = True
says_you_won = True

# The email looks like spam only if ALL three clues are present
looks_like_spam = has_link and from_unknown_sender and says_you_won
print("Spam?", looks_like_spam)
````

> **What happened?** All three clues were True, so the result is True. Real AI does **not** use fixed rules like this - it **learns** which clues matter from thousands of labelled emails. But its final answer is still a True/False value like this one!

## 11. Neat output with f-strings
An **f-string** is the easiest way to mix text and variables. Write `f` before the opening quote and put variable names inside curly brackets `{ }`.
- `{average:.2f}` shows the value with **2 decimal places**.
- You can even write a calculation inside `{ }`.

````python
name = "Ravi"
total = 437
subjects = 5
average = total / subjects           # 87.4

print(f"{name} scored {total} marks in {subjects} subjects.")   # values go into the {}
print(f"Average = {average}")
print(f"Average (2 decimals) = {average:.2f}")                   # :.2f -> 2 decimal places
print(f"Percentage = {total / 500 * 100:.1f}%")                  # calculation inside {}
````

> **What happened?** Python replaced each `{ }` with its value. f-strings are cleaner than joining text with `+` and `str()`, and you will use them in almost every program.

## 12. Common mistakes and error messages
Errors are normal - every programmer sees them every day. Read the **last line** of the error message: it tells you the type of error and what went wrong.

| Error | Example | Why | Fix |
|---|---|---|---|
| `SyntaxError` | `print("hi"` | missing bracket or quote | close it: `print("hi")` |
| `NameError` | `print(nmae)` | spelling mistake, or variable not created yet | check spelling; run the cell that creates it |
| `TypeError` | `"Age: " + 18` | cannot join text and a number | `"Age: " + str(18)` or use an f-string |
| `ValueError` | `int("abc")` | the text is not a number | check the input |
| `ZeroDivisionError` | `10 / 0` | cannot divide by zero | check the divisor first |

**Activity:** the cell below has a mistake **on purpose**. Run it, read the red error message, and then fix the code.

````python
age = 18
print("Age: ",+ age)    # this line has a mistake - can you fix it?
````

> **What happened?** You should see `TypeError: can only concatenate str (not "int") to str`. Python cannot join the text `"Age: "` with the number `18`. Two correct fixes: `print("Age: " + str(age))` or `print(f"Age: {age}")`.

## Mini project: Mess bill calculator
**Mini project.** Write a program that reads:
- number of days a student ate in the mess this month
- cost per day (Rs.)
- number of special dinners (Rs. 60 extra each)

Print a neat bill showing the **food cost**, **special dinner cost**, **5% GST** on the total, and the **grand total**, all with 2 decimal places.

<details><summary><b>Hint</b> (click to open)</summary>

Work in three steps: (1) read the inputs, (2) calculate food, special cost, subtotal, GST = subtotal × 5 / 100, grand total, (3) print with f-strings and `:.2f`.

</details>

````python
days = int(input("Enter number of mess days: "))
cost_per_day = float(input("Enter cost per day: "))
special_dinners = int(input("Enter number of special dinners: "))

food_cost = days * cost_per_day
special_cost = special_dinners * 60

subtotal = food_cost + special_cost
gst = subtotal * 0.05
grand_total = subtotal + gst

print("\n========== MESS BILL ==========")
print(f"Food Cost: Rs. {food_cost:.2f}")
print(f"Special Dinner Cost: Rs. {special_cost:.2f}")
print(f"GST (5%): Rs. {gst:.2f}")
print("-------------------------------")
print(f"Grand Total: Rs. {grand_total:.2f}")
````
