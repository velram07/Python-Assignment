## 1. Strings in depth
A string is a **sequence of characters**. Each character has a position number called an **index**.
- Indexes start from **0**, not 1.
- **Negative** indexes count from the end: `-1` is the last character.

```
 P   Y   T   H   O   N
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```
Use square brackets to get one character: `word[0]`.

````python
word = "PYTHON"
print(word[0])      # index 0 -> first character 'P'
print(word[3])      # index 3 -> fourth character 'H'
print(word[-1])     # index -1 -> last character 'N'
print(len(word))    # len() gives the number of characters -> 6
````

> **What happened?** `word[3]` is **H**, the fourth letter, because counting starts at 0. The last valid index is `len(word) - 1` = 5. Asking for `word[6]` would give an `IndexError`.

### Slicing: `text[start:stop]`
A **slice** takes a part of a string. It starts at `start` and stops **before** `stop` (just like `range`).
- Leave out `start` to begin at the start; leave out `stop` to go to the end.
- `[::-1]` uses a step of -1, which **reverses** the string.

````python
dept = "Artificial Intelligence"
print(dept[0:10])     # characters 0 to 9 -> 'Artificial'
print(dept[11:])      # from index 11 to the end -> 'Intelligence'
print(dept[:3])       # from the start up to index 2 -> 'Art'
print(dept[::-1])     # step -1 -> the whole string reversed
````

> **What happened?** `dept[0:10]` gave the 10 characters at positions 0-9. Index 10 is the space, so `Intelligence` starts at index 11. The reversed string is useful for checking palindromes (see the exercises).

### Useful string methods
A **method** is a function that belongs to a value. Call it with a dot: `text.upper()`.
String methods **return a new string** - they do not change the original.

| Method | What it does |
|---|---|
| `strip()` | removes spaces at both ends |
| `upper()` / `lower()` / `title()` | CAPITALS / small letters / Each Word Capitalised |
| `replace(a, b)` | replaces `a` with `b` |
| `find(x)` | position of `x` (-1 if not found) |
| `count(x)` | how many times `x` appears |
| `startswith(x)` / `endswith(x)` | True/False |

````python
name = "  asha kumar  "
print(name.strip())            # 'asha kumar' - spaces at both ends removed
print(name.strip().title())    # methods can be chained: strip, then title -> 'Asha Kumar'
print(name.upper())
print(name.lower())

email = "asha.k@bitsathy.ac.in"
print(email.endswith(".ac.in"))               # does it end with .ac.in? -> True
print(email.find("@"))                        # position of '@' -> 6
print(email.count("a"))                       # how many 'a' characters
print(email.replace("asha.k", "ravi.m"))      # a new string with the name replaced
````

> **What happened?** Chaining `.strip().title()` first removed the spaces and then capitalised each word. `find("@")` returned 6 - the index where `@` is. Cleaning text like this is the first step in every language-AI (NLP) project.

### `split()` and `join()`
- `split()` breaks a string into a **list** of pieces. With no argument it splits at spaces; `split(",")` splits at commas.
- `join()` does the opposite: it glues a list of strings into one string, with the chosen text between the pieces.

````python
sentence = "AI learns patterns from data"
words = sentence.split()             # split at spaces -> a list of 5 words
print(words)
print("Number of words:", len(words))

print("-".join(words))               # glue the words together with '-' between them

marks_line = "78,85,92,66"
print(marks_line.split(","))         # split at commas -> ['78', '85', '92', '66']
````

> **What happened?** `split()` turned one sentence into a list of words, which is how a computer starts to 'read' text. Note that `"78,85,92,66".split(",")` gives **strings** like `'78'` - convert them with `int()` before doing maths.

### Strings cannot be changed in place
Strings are **immutable** (unchangeable). `word[0] = "J"` gives a `TypeError`. To 'change' a string, build a **new** one from pieces.

````python
word = "Python"
new_word = "J" + word[1:]     # 'J' + 'ython' -> a brand-new string
print(new_word)
print(word)                   # the original is unchanged
````

> **What happened?** We took the slice `word[1:]` (everything except the first letter) and joined `J` in front. The original `word` is still `Python`.

## 2. Lists: many values in one variable
Imagine storing marks of 60 students in 60 separate variables - impossible! A **list** stores many values in **one** variable.
- Write a list with square brackets `[ ]` and commas between items.
- Items keep their **order** and have indexes (from 0), just like strings.
- Lists are **mutable**: you can add, remove and change items.

````python
marks = [78, 85, 92, 66, 70]
subjects = ["Maths", "Physics", "Chemistry", "Python", "English"]

print(marks)
print(subjects[0])        # first item
print(subjects[-1])       # last item
print(marks[1:3])         # slice: items at index 1 and 2 -> [85, 92]
print(len(marks))         # number of items -> 5
````

> **What happened?** Indexing and slicing work exactly as with strings. A slice of a list gives another list.

### Changing a list
| Method | What it does |
|---|---|
| `append(x)` | add `x` at the end |
| `insert(i, x)` | add `x` at position `i` |
| `list[i] = x` | replace the item at position `i` |
| `remove(x)` | remove the first `x` found |
| `pop()` | remove the last item and give it back |

````python
fruits = ["apple", "banana"]
fruits.append("mango")          # add at the end
fruits.insert(1, "orange")      # add at position 1; others move right
print(fruits)

fruits[0] = "green apple"       # replace the item at index 0
print(fruits)

fruits.remove("banana")         # remove by value
last = fruits.pop()             # remove the last item and store it in 'last'
print("Removed:", last)
print(fruits)
````

> **What happened?** Unlike strings, the **same** list changed after each step. Watch how `insert(1, ...)` pushed `banana` from index 1 to index 2.

### Built-in functions and sorting
Python has functions that work on a whole list at once: `sum()`, `max()`, `min()`, `len()`.
- `list.sort()` sorts the list **itself** (changes it).
- `sorted(list)` gives a **new** sorted list and leaves the original unchanged.
- `x in list` checks whether `x` is in the list (True/False).

````python
marks = [78, 85, 92, 66, 70]
print("Total:", sum(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print(f"Average: {sum(marks) / len(marks):.2f}")   # total divided by count

marks.sort()                       # sort the list itself, smallest first
print("Ascending:", marks)
marks.sort(reverse=True)           # largest first
print("Descending:", marks)

print(sorted(["Ravi", "Asha", "Kiran"]))   # a new list in alphabetical order
print(85 in marks)                         # is 85 in the list? -> True
````

> **What happened?** One line each gave the total, highest, lowest and average - compare this with the loop patterns of Notebook 2. Average = `sum / len` is the most used formula in data analysis.

### Looping over lists
- `for item in list:` - visit each item.
- `enumerate()` - gives the **position and the item** together (we start counting at 1 here).
- `zip()` - walks through **two lists side by side**.

````python
subjects = ["Maths", "Physics", "Python"]
marks = [88, 74, 95]

for s in subjects:                          # s takes each subject in turn
    print("Subject:", s)

for i, s in enumerate(subjects, start=1):   # i = 1, 2, 3 and s = each subject
    print(i, s)

for s, m in zip(subjects, marks):           # pairs: (Maths, 88), (Physics, 74), (Python, 95)
    print(f"{s:10} {m}")                    # {s:10} uses 10 spaces so the columns line up
````

> **What happened?** `zip()` matched each subject with its mark by position. This is how you compare actual and predicted labels in AI - item 1 with item 1, item 2 with item 2, and so on.

### Building a list with a loop
Start with an **empty list** `[]` and `append()` a new item in each round of the loop.

````python
squares = []                    # empty list
for n in range(1, 6):
    squares.append(n * n)       # add the square of n to the list
print(squares)
````

> **What happened?** The list grew from `[]` to `[1, 4, 9, 16, 25]`, one item per round.

### List comprehension (a short way to build lists)
Python lets you build a list in **one line**:

`[expression for item in sequence if condition]`

Read it as: *"make a list of `expression`, for each `item` in `sequence`, but only if `condition` is True"*. The `if` part is optional.

````python
squares = [n * n for n in range(1, 6)]            # same result as the loop above
print(squares)

marks = [78, 45, 92, 38, 85]
passed = [m for m in marks if m >= 50]            # keep only the marks >= 50
print("Passed marks:", passed)

names = ["asha", "ravi", "kiran"]
print([n.title() for n in names])                 # change every item
````

> **What happened?** Each comprehension replaced a 3-4 line loop. Use them for simple cases; for complicated logic a normal loop is easier to read.

## 3. Tuples: lists that cannot change
A **tuple** is like a list but written with round brackets `( )` and it **cannot be changed** after it is created.
Use it for values that belong together and should stay fixed, such as a date or a location.

**Unpacking** puts each item of a tuple into its own variable in one line.

````python
campus_location = (11.4966, 77.2770)   # (latitude, longitude) - approximate
dob = (15, 8, 2008)                    # (day, month, year)

print(campus_location[0])              # index works like a list
day, month, year = dob                 # unpacking: 3 items into 3 variables
print(f"Born on {day}/{month}/{year}")
````

> **What happened?** Reading items works the same as a list, but `dob[0] = 20` would give an error. That protection is the reason to use a tuple for fixed data.

## 4. Dictionaries: data with labels
A list finds items by **position** (0, 1, 2...). A **dictionary** finds items by a **name** called a **key**.
It stores **key: value** pairs inside curly brackets `{ }`, just like a real dictionary stores *word: meaning*.

Use `dict[key]` to get a value. Use `dict.get(key, default)` when the key might be missing - it returns the default instead of an error.

````python
student = {
    "name": "Asha",       # key "name"  -> value "Asha"
    "roll": 101,
    "dept": "AI and DS",
    "cgpa": 8.6,
}

print(student["name"])                      # look up by key
print(student["cgpa"])
print(student.get("phone", "not given"))    # key missing -> returns "not given"
````

> **What happened?** `student["name"]` is much clearer than `student[0]` - you do not have to remember positions. `student["phone"]` would give a `KeyError`, but `.get()` returned the default safely.

### Adding, changing and removing
- `dict[key] = value` **adds** a new pair if the key is new, or **changes** the value if the key already exists.
- `del dict[key]` removes a pair.
- `key in dict` checks whether a key exists.

````python
student = {"name": "Asha", "roll": 101}
student["hostel"] = True        # new key -> added
student["roll"] = 102           # existing key -> value changed
print(student)

del student["hostel"]           # remove the pair with key "hostel"
print(student)
print("name" in student)        # does the key "name" exist? -> True
````

> **What happened?** The same syntax `student[key] = value` both added `hostel` and updated `roll`. Keys are unique - a dictionary cannot have two `roll` keys.

### Looping over a dictionary
- `for key in dict:` gives each **key**.
- `for key, value in dict.items():` gives each **key and value** together - the most useful form.
- `.keys()` and `.values()` give all keys or all values.

````python
marks = {"Maths": 88, "Physics": 74, "Python": 95}

for subject in marks:                    # loops over the keys
    print(subject)

for subject, mark in marks.items():      # key and value together
    print(f"{subject}: {mark}")

print("Keys:", list(marks.keys()))
print("Values:", list(marks.values()))
print("Total:", sum(marks.values()))     # add up all the values
````

> **What happened?** `.items()` gave each subject with its mark. `sum(marks.values())` added only the numbers, giving 257.

### AI connection: counting words
Spam filters and chatbots begin by **counting words**. A dictionary is perfect for this: the **word** is the key and its **count** is the value.
For each word: if it is already in the dictionary, add 1; otherwise, start its count at 1.

````python
message = "win win free money click now win free gift"
counts = {}                          # empty dictionary
for word in message.split():         # go through each word
    if word in counts:
        counts[word] += 1            # seen before -> increase its count
    else:
        counts[word] = 1             # first time -> start at 1
print(counts)
print("Most common word:", max(counts, key=counts.get))   # key with the biggest value
````

> **What happened?** `win` appears 3 times and `free` twice. Turning text into word counts like this is called a **bag of words** - one of the oldest ways to feed text into a machine learning model. (`max(counts, key=counts.get)` finds the key whose value is largest.)

## 5. Sets: unique items only
A **set** is written with `{ }` (without key: value pairs). It has two special properties:
- **No duplicates** - each item appears only once.
- **No order** - items have no index, and the printed order may vary.

Sets are great for removing repeats and for comparing groups:
`&` items in **both**, `|` items in **either**, `-` items in the first but **not** the second.

````python
visitors = ["Asha", "Ravi", "Asha", "Kiran", "Ravi"]
unique = set(visitors)                 # duplicates removed automatically
print(unique)
print("Unique visitors:", len(unique))

asha_courses = {"Python", "AI", "Maths", "English"}
ravi_courses = {"Python", "Physics", "Maths"}

print("Both:", asha_courses & ravi_courses)       # intersection - in both sets
print("Either:", asha_courses | ravi_courses)     # union - in at least one set
print("Only Asha:", asha_courses - ravi_courses)  # difference - in Asha's but not Ravi's
````

> **What happened?** Five names became 3 unique visitors. The set operations answered *who takes which course* in one line each. Do not worry if the order of items is different on your screen - sets have no order.

## 6. Which collection should I use?
| Type | Brackets | Ordered? | Changeable? | Duplicates? | Use it for |
|---|---|---|---|---|---|
| list | `[ ]` | yes | yes | yes | marks of a class, a to-do list |
| tuple | `( )` | yes | no | yes | a date, coordinates |
| dict | `{key: value}` | yes | yes | keys unique | a student record, word counts |
| set | `{ }` | no | yes | no | unique items, common items |

## 7. A small dataset: list of dictionaries
In Week 3 you stored class data in a spreadsheet: **each row is a record, each column is a feature**.
In Python, the same table is a **list of dictionaries**: each dictionary is one **row**, and its keys are the **column names**.
The cell below stores 5 rows and prints them as a neat table.

````python
students = [
    {"roll": 101, "name": "Asha",  "type": "Hostel",      "sleep": 7.0, "marks": 86},
    {"roll": 102, "name": "Ravi",  "type": "Day scholar", "sleep": 6.0, "marks": 72},
    {"roll": 103, "name": "Kiran", "type": "Day scholar", "sleep": 6.5, "marks": 64},
    {"roll": 104, "name": "Meena", "type": "Hostel",      "sleep": 8.0, "marks": 91},
    {"roll": 105, "name": "Arun",  "type": "Hostel",      "sleep": 5.5, "marks": 58},
]

# Header row. :<6 means 'left-align in 6 spaces', :>6 means 'right-align in 6 spaces'
print(f"{'Roll':<6}{'Name':<8}{'Type':<13}{'Sleep':>6}{'Marks':>7}")
for s in students:                     # s is one row (one dictionary)
    print(f"{s['roll']:<6}{s['name']:<8}{s['type']:<13}{s['sleep']:>6}{s['marks']:>7}")
````

> **What happened?** Each loop round took one dictionary (row) and printed its values in fixed-width columns. Inside an f-string we used single quotes `s['name']`, because the f-string itself uses double quotes.

Now we can **ask questions** about the data with the tools from this notebook: a comprehension to filter rows, `sum()/len()` for an average, and the *find the largest* pattern for the topper.

````python
# 1. Filter: keep only the rows where type is "Hostel"
hostel = [s for s in students if s["type"] == "Hostel"]
print("Hostel students:", [s["name"] for s in hostel])

# 2. Average of one column for the filtered rows
avg_sleep = sum(s["sleep"] for s in hostel) / len(hostel)
print(f"Average sleep of hostel students: {avg_sleep:.1f} hours")

# 3. Find the row with the highest marks
top = students[0]                    # assume the first student is the topper
for s in students:
    if s["marks"] > top["marks"]:    # found a higher mark?
        top = s
print("Topper:", top["name"], top["marks"])
````

> **What happened?** Three hostel students sleep 6.8 hours on average, and Meena is the topper. Next semester you will use **Pandas**, which answers questions like these on thousands of rows in one line. Now you know what happens underneath!

### Exercise 10: Simple spam checker (AI from Zero link)
Create a **set** of spam words: `{"win", "free", "prize", "click", "offer", "lottery"}`. Read a message, count how many of its words are spam words, and print `SPAM` if there are **2 or more**, otherwise `NOT SPAM`.

Split the lower-case message into words. Keep the words that are `in spam_words`, then check the length of that list.

results = {"Asha": [88, 92, 79], "Ravi": [45, 60, 52], "Kiran": [35, 48, 41], "Meena": [95, 89, 97], "Arun": [70, 66, 58]}

For each student print the **total**, **average** and **result** (`Pass` if every subject is 50 or more, else `Fail`). Then print the **class topper** and the **pass percentage** of the class.

Loop with `for name, marks in results.items():`. A student passes if `min(marks) >= 50`. Use a counter for passes and the find-the-largest pattern for the topper.

</details>

````python
results = {
    "Asha": [88, 92, 79],
    "Ravi": [45, 60, 52],
    "Kiran": [35, 48, 41],
    "Meena": [95, 89, 97],
    "Arun": [70, 66, 58]
}

topper = ""
highest_average = -1
pass_count = 0

print("STUDENT RESULT REPORT")
print("-" * 60)

for name, marks in results.items():
    total = sum(marks)
    average = total / len(marks)

    if all(mark >= 50 for mark in marks):
        result = "Pass"
        pass_count += 1
    else:
        result = "Fail"

    print(f"{name}: Total={total}, Average={average:.2f}, Result={result}")

    if average > highest_average:
        highest_average = average
        topper = name

pass_percentage = (pass_count / len(results)) * 100

print("-" * 60)
print("Class Topper:", topper)
print("Pass Percentage:", f"{pass_percentage:.2f}%")
````
