## 1. Why functions?
A **function** is a named block of code that does one job. You **define** it once and then **call** it (use it) as many times as you like.
You have already used built-in functions: `print()`, `len()`, `input()`, `sum()`, `max()`. Now you will write your own.

**Why bother?**
- No repeated code - write once, use many times.
- Easier to read - a good name like `calculate_gpa()` explains what the code does.
- Easier to fix - a bug is fixed in one place only.

**Syntax:**
```
def function_name():      # def = define; ends with a colon
    indented lines        # the body of the function
```

````python
def greet():                                          # define a function called greet
    print("Welcome to the Python lab!")               # the body (indented)
    print("Please log in with your roll number.")

greet()      # call the function - its body runs now
greet()      # call it again - the body runs again
````

> **What happened?** Defining the function printed **nothing** - `def` only stores the instructions. The messages appeared only when we **called** `greet()`. Each call ran the whole body, so we got the two lines twice.

### Parameters: giving input to a function
A function becomes more useful when it can work with different data. Names inside the brackets of `def` are **parameters** - placeholders for values.
The actual values you pass when calling are called **arguments**.

````python
def greet(name, dept):                       # name and dept are parameters
    print(f"Hello {name} from {dept}!")

greet("Asha", "AI and DS")                   # arguments: name = "Asha", dept = "AI and DS"
greet("Ravi", "Mechanical")                  # same function, different data
````

> **What happened?** In the first call, `name` held "Asha"; in the second, "Ravi". Arguments are matched to parameters **by position**: first to first, second to second.

### `return`: giving a result back
`print()` only **shows** a value on the screen. `return` **sends the value back** to the place where the function was called, so you can store it in a variable, use it in a calculation, or check it with `if`.
When Python reaches `return`, the function ends immediately.

````python
def average(a, b, c):
    return (a + b + c) / 3          # send the result back to the caller

avg = average(80, 90, 70)           # the returned value (80.0) is stored in avg
print("Average:", avg)

if average(40, 45, 50) < 50:        # the returned value can be used directly in a condition
    print("Needs improvement")
````

> **What happened?** `average(80, 90, 70)` was replaced by its result, 80.0, which we stored in `avg`. In the `if`, the result 45.0 was compared with 50 straight away.

### `print` vs `return` - the most important difference
Run the cell and compare the values of `x` and `y`.
A function with no `return` gives back the special value `None` (*nothing*).

````python
def add_print(a, b):
    print(a + b)          # only SHOWS the answer

def add_return(a, b):
    return a + b          # GIVES BACK the answer

x = add_print(2, 3)       # shows 5 on screen, but x receives nothing (None)
y = add_return(2, 3)      # shows nothing, but y receives 5
print("x =", x, " y =", y)
````

> **What happened?** `x` is `None` because `add_print` never returned anything - it only printed. `y` is 5. **Rule of thumb:** if you need to use the answer later, `return` it.

### Functions that make decisions
A function can contain `if/elif/else`. Each branch can `return` a different value. Once a `return` runs, the function stops.

````python
def grade(marks):
    if marks >= 90:
        return "O"          # function ends here for marks >= 90
    elif marks >= 80:
        return "A+"
    elif marks >= 70:
        return "A"
    elif marks >= 60:
        return "B+"
    elif marks >= 50:
        return "B"
    else:
        return "RA"

for m in [95, 83, 71, 49]:      # call the same function for four marks
    print(m, "->", grade(m))
````

> **What happened?** The grade logic from Notebook 2 is now packed into a reusable function. Any program can call `grade()` without knowing how it works inside.

### Default values and keyword arguments
- A **default value** (`cost_per_day=120`) is used when the caller does not give that argument.
- A **keyword argument** (`gst=0`) names the parameter when calling, so the order does not matter.

````python
def mess_bill(days, cost_per_day=120, gst=5):     # two parameters have defaults
    subtotal = days * cost_per_day
    return subtotal + subtotal * gst / 100

print(mess_bill(30))                  # uses cost_per_day=120 and gst=5
print(mess_bill(30, 150))             # cost_per_day=150, gst still 5
print(mess_bill(days=25, gst=0))      # keyword arguments: cost stays 120, gst becomes 0
````

> **What happened?** Call 1: 30 × 120 + 5% = 3780.0. Call 2: 30 × 150 + 5% = 4725.0. Call 3: 25 × 120 with no GST = 3000.0. Defaults make functions easy to use for the common case but still flexible.

### Returning more than one value
A function can return several values separated by commas. The caller **unpacks** them into variables (like tuple unpacking in Notebook 3).

````python
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)   # three values

low, high, avg = stats([56, 78, 91, 45, 67])     # unpack into three variables
print(f"Lowest {low}, highest {high}, average {avg:.1f}")
````

> **What happened?** Python packed the three values into a tuple and the caller unpacked them in order: `low` = 45, `high` = 91, `avg` = 67.4.

### Docstrings
A **docstring** is a string in triple quotes written just below `def`. It explains what the function does. `help(function_name)` shows it.
Good programmers write one for every function.

````python
def accuracy(correct, total):
    """Return the accuracy as a percentage, e.g. accuracy(9, 10) -> 90.0"""
    return correct / total * 100

print(accuracy(9, 10))
help(accuracy)          # displays the docstring
````

> **What happened?** `help()` printed the function's name, parameters and docstring. In Colab you can also hover over a function name to see its docstring.

### Local and global variables
- A variable created **outside** all functions is **global** - it can be read everywhere.
- A variable created **inside** a function is **local** - it exists only while that function runs and cannot be used outside.

This keeps functions independent: a variable named `total` inside one function does not disturb a `total` elsewhere.

````python
college = "BIT"          # global variable

def show():
    year = 1             # local variable - lives only inside show()
    print(college, "year", year)   # a function can read a global variable

show()
print(college)           # works: college is global
# print(year)            # remove the # and run: NameError, because year is local to show()
````

> **What happened?** `show()` could use both `college` and `year`. Outside the function only `college` exists. Try removing the `#` on the last line to see the `NameError` for yourself.

### Extra: `lambda` (small one-line functions)
A `lambda` is a tiny function without a name, written in one line: `lambda parameters: expression`.
You will often see it in AI code, mostly to tell `sort()` or `max()` **what to compare**.

````python
square = lambda x: x * x                  # same as: def square(x): return x * x
print(square(6))

students = [("Asha", 86), ("Ravi", 72), ("Meena", 91)]
students.sort(key=lambda s: s[1], reverse=True)    # sort by item [1] (the marks), largest first
print(students)
````

> **What happened?** `key=lambda s: s[1]` told `sort()` to compare the students by their marks (index 1 of each tuple) instead of by their names. `lambda` is optional for this course - a normal `def` function always works too.

## 2. Modules: ready-made code
A **module** is a file of useful functions written by other people. Python comes with hundreds of them. Use `import` to bring a module in, then use its functions with a dot: `module.function()`.

| Module | Useful for |
|---|---|
| `math` | square root, pi, rounding up/down, factorial |
| `random` | random numbers, random choices, shuffling |
| `datetime` | today's date, differences between dates |

First, the `math` module:

````python
import math                           # load the math module

print(math.sqrt(144))                 # square root -> 12.0
print(math.pi)                        # the value of pi
print(math.ceil(4.1), math.floor(4.9))  # round UP -> 5, round DOWN -> 4
print(math.factorial(5))              # 5! -> 120
````

> **What happened?** We did not write any of these functions - we just used them. `math.factorial(5)` gives the same 120 that you calculated with a loop in Notebook 2.

The `random` module makes programs unpredictable - useful for games, quizzes and for **shuffling data** before training an AI model. Run the cell several times: the output changes each time.

````python
import random

print(random.randint(1, 6))                          # random whole number from 1 to 6 (a dice)
print(random.choice(["rock", "paper", "scissors"]))  # pick one item from a list

names = ["Asha", "Ravi", "Kiran", "Meena"]
random.shuffle(names)                                # mix the list into a random order
print(names)
````

> **What happened?** Every run gives different results. In machine learning, data is shuffled before splitting it into training and test sets, so that the split is fair.

The `datetime` module works with dates. Subtracting two dates gives the number of days between them. Here we use `from ... import ...` to bring in only `date`.

````python
from datetime import date            # import only the 'date' part of the module

today = date.today()                 # today's date from the computer's clock
print("Today:", today)
exam = date(2027, 4, 20)             # a sample exam date: year, month, day
print("Days left for the exam:", (exam - today).days)   # difference in days
````

> **What happened?** `exam - today` gives a time difference, and `.days` turns it into a whole number of days. Your answer depends on the day you run it.

## 3. Working with files
Data in variables **disappears** when the program ends. **Files** keep data saved for later.

To use a file: **open** it, **read or write**, then **close** it. `with open(...) as f:` does all of this and closes the file automatically, even if an error happens.

| Mode | Meaning |
|---|---|
| `"w"` | **write** - creates a new file; **erases** old content if the file exists |
| `"a"` | **append** - adds to the end of the file |
| `"r"` | **read** - the default |

`"\n"` is the **new line** character. `write()` does not add it automatically.

*Colab note: files are saved in the Colab session (click the folder icon on the left to see them). They are deleted when the session ends - download them if you need them.*

````python
with open("marks.txt", "w") as f:    # open marks.txt for writing; f is our name for the file
    f.write("Asha,86\n")             # \n moves to a new line in the file
    f.write("Ravi,72\n")
    f.write("Meena,91\n")
print("File saved")                  # this line is outside 'with' - the file is closed now
````

> **What happened?** A file called `marks.txt` now exists with 3 lines. Open the folder icon on the left of Colab to see it (you may need to click refresh).

Now we **append** one more line (mode `"a"` keeps the old content), and then **read** the whole file with `read()`.

````python
with open("marks.txt", "a") as f:    # append mode - add to the end, keep old lines
    f.write("Kiran,64\n")

with open("marks.txt", "r") as f:    # read mode
    content = f.read()               # read the whole file into one string
print(content)
````

> **What happened?** All 4 lines were printed: the 3 old ones plus Kiran. If we had used `"w"` instead of `"a"`, the file would contain only Kiran.

### Reading line by line and using the data
A `for` loop over a file gives **one line at a time**. For each line we:
1. `strip()` the `\n` at the end,
2. `split(",")` it into name and mark,
3. convert the mark to `int` so we can calculate.

````python
total = 0
count = 0
with open("marks.txt", "r") as f:
    for line in f:                              # one line at a time, e.g. "Asha,86\n"
        name, mark = line.strip().split(",")    # -> "Asha" and "86"
        mark = int(mark)                        # text -> number
        total += mark                           # accumulator
        count += 1                              # counter
        print(f"{name:<6} {mark}")
print(f"Class average: {total / count:.2f}")
````

> **What happened?** The program read the data back from the file and calculated the class average, 78.25. A file whose values are separated by commas is called a **CSV** (comma-separated values) file. Google Sheets can save data as CSV, and most AI datasets are CSV files.

## 4. Handling errors with `try` and `except`
When something goes wrong (the user types letters instead of a number, a file is missing...), Python stops with an error - the program **crashes**.
`try ... except` lets the program **catch** the error and handle it politely:
```
try:
    code that might fail
except ErrorType:
    what to do if that error happens
```
Run the cell below **twice**: first type `18`, then type `eighteen`.

````python
try:
    age = int(input("Enter your age: "))     # int() fails if the text is not a number
    print("Next year you will be", age + 1)
except ValueError:                           # runs only if a ValueError happened above
    print("Please enter a number, like 18")
````

> **What happened?** With `18`, the `try` block finished normally and the `except` block was skipped. With `eighteen`, `int()` raised a `ValueError`; Python jumped to `except` and printed a friendly message instead of crashing.

Different problems raise different errors. Here we catch `ZeroDivisionError` inside a function and return `None` (*no result*) instead of crashing.

````python
def safe_divide(a, b):
    try:
        return a / b                       # may fail if b is 0
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None                        # no valid result

print(safe_divide(10, 2))                  # works normally -> 5.0
print(safe_divide(10, 0))                  # handled -> message, then None
````

> **What happened?** The first call returned 5.0. The second call hit the error, printed the message and returned `None`, and the program **kept running**.

Opening a file that does not exist raises `FileNotFoundError`.
`finally` is an optional block that **always** runs, whether there was an error or not - useful for clean-up messages.

````python
try:
    with open("attendance.txt", "r") as f:   # this file was never created
        print(f.read())
except FileNotFoundError:
    print("attendance.txt does not exist yet")
finally:
    print("This line always runs")
````

> **What happened?** Because the file is missing, the `except` block ran, and then `finally` ran as well.

### Keep asking until the input is valid
Combine a `while True` loop (repeat forever) with `try/except`. The loop only ends when `return` gives back a valid number.
This function is a handy tool you can reuse in any program that reads marks.

````python
def read_marks(prompt):
    while True:                                  # keep asking...
        try:
            m = int(input(prompt))
            if 0 <= m <= 100:                    # valid range?
                return m                         # ...until we get a good value (return ends the loop)
            print("Marks must be between 0 and 100")
        except ValueError:
            print("That is not a number. Try again.")

marks = read_marks("Enter marks (0-100): ")
print("You entered", marks)
````

> **What happened?** Try typing `abc`, then `150`, then `75`. The first two are rejected with a message, and the loop asks again. Only a valid mark is returned.

## 5. AI connection: a tiny 'model' as a function
A trained AI model behaves like a function: **features go in, a prediction comes out**.
Below, a **rule-based** function predicts whether a banana is ripe from two features (how yellow it is, and how soft it is, on a 0-100 scale). Then we test it on 5 labelled examples and calculate the accuracy - just like testing your Teachable Machine model.

````python
def predict_ripe(yellow_percent, softness):
    """Predict 'ripe' or 'unripe' for a banana from two features (0-100 scale)."""
    if yellow_percent >= 70 and softness >= 40:     # our hand-written rule
        return "ripe"
    return "unripe"

# Test data: (yellow %, softness, actual label)
test_data = [(90, 60, "ripe"), (40, 20, "unripe"), (75, 30, "ripe"), (95, 80, "ripe"), (30, 50, "unripe")]

correct = 0
for yellow, soft, actual in test_data:              # unpack each tuple
    pred = predict_ripe(yellow, soft)               # the 'model' makes a prediction
    if pred == actual:
        correct += 1
        mark = "correct"
    else:
        mark = "WRONG"
    print(f"yellow={yellow:3} soft={soft:3} -> predicted {pred:7} actual {actual:7} {mark}")

print(f"Accuracy: {correct / len(test_data) * 100:.0f}%")
````

> **What happened?** The third banana (yellow 75, softness 30) is actually ripe, but our rule said unripe, so accuracy is 80%. How would you change the rule? Trying rules, measuring accuracy and adjusting is exactly what **training** does - automatically, with thousands of examples.

## 6. Putting it together: Student Result Manager
This program uses **everything** from the four notebooks: functions, a dictionary of lists, loops, `if`, f-strings, files and `try/except`.
Read the docstring and comments of each function, then run the cell.

| Function | Job |
|---|---|
| `add_student()` | add a student and their marks to the dictionary |
| `result()` | decide Pass or Fail |
| `save()` | write all students to a CSV file |
| `load()` | read the CSV file back into a dictionary (safely) |
| `report()` | print a neat result table |

````python
def add_student(db, name, marks):
    """Add a student and their list of marks to the database dictionary."""
    db[name] = marks

def result(marks):
    """Return 'Pass' if every mark is 50 or more, otherwise 'Fail'."""
    return "Pass" if min(marks) >= 50 else "Fail"

def save(db, filename):
    """Write each student as one line: name,mark1,mark2,..."""
    with open(filename, "w") as f:
        for name, marks in db.items():
            f.write(name + "," + ",".join(str(m) for m in marks) + "\n")

def load(filename):
    """Read the file back into a dictionary. Start empty if the file is missing."""
    db = {}
    try:
        with open(filename) as f:
            for line in f:
                parts = line.strip().split(",")          # ['Asha', '88', '92', '79']
                db[parts[0]] = [int(x) for x in parts[1:]]   # name -> list of int marks
    except FileNotFoundError:
        print("No saved file found - starting fresh")
    return db

def report(db):
    """Print total, average and result for every student."""
    print(f"{'Name':<8}{'Total':>6}{'Avg':>7}  Result")
    for name, marks in db.items():
        print(f"{name:<8}{sum(marks):>6}{sum(marks) / len(marks):>7.1f}  {result(marks)}")

# ---- main program: uses the functions above ----
students = {}
add_student(students, "Asha", [88, 92, 79])
add_student(students, "Ravi", [45, 60, 52])
add_student(students, "Meena", [95, 89, 97])
save(students, "results.csv")       # store on disk

loaded = load("results.csv")        # read back from disk
report(loaded)
````

> **What happened?** The data was saved to `results.csv`, read back, and reported. Because each job is in its own small function, the **main program** at the bottom reads almost like English. This is how real software - and real AI pipelines - are organised.

### Exercise 10: Accuracy function (AI from Zero link)
Write a function `accuracy(actual, predicted)` that takes two lists of labels and returns the accuracy in %. Test it with:
```python
actual    = ["rock", "paper", "scissors", "rock", "paper"]
predicted = ["rock", "paper", "rock", "rock", "scissors"]
```
*Expected:* 60.0

<details><summary><b>Hint</b> (click to open)</summary>

Use `zip(actual, predicted)` to compare item by item and count the matches.

</details>

````python
def accuracy(actual, predicted):
    correct = 0

    for a, p in zip(actual, predicted):
        if a == p:
            correct += 1

    return (correct / len(actual)) * 100

actual = ["rock", "paper", "scissors", "rock", "paper"]

predicted = ["rock", "paper", "rock", "rock", "scissors"]

print("Accuracy:", accuracy(actual, predicted))
````

## Final mini project: Quiz app
**Final mini project.** Build a small quiz app:
1. Store 5 questions and answers in a **dictionary** (for example AI from Zero questions).
2. Write a function `ask(question, answer)` that asks the question with `input()`, compares the answer (ignore capitals and spaces) and returns `True` or `False`.
3. Loop through all questions and count the score.
4. Save the player's name and score to `scores.txt` using append mode.
5. Print the score and a message: 5 = *Excellent*, 3-4 = *Good*, below 3 = *Keep practising*.

<details><summary><b>Hint</b> (click to open)</summary>

Store questions as keys and answers (in lower case) as values. In `ask()`, return `reply.strip().lower() == answer`. Loop with `.items()`, count the score, then write to the file with mode `"a"`.

</details>

````python
questions = {
    "What does AI stand for?": "artificial intelligence",
    "Which language are we learning?": "python",
    "What does CPU stand for?": "central processing unit",
    "Which data type stores True or False?": "bool",
    "What function displays output in Python?": "print"
}

def ask(question, answer):
    reply = input(question + " ")
    return reply.strip().lower() == answer.strip().lower()

name = input("Enter your name: ")

score = 0

print("\nWELCOME TO THE PYTHON QUIZ!")

for question, answer in questions.items():
    if ask(question, answer):
        print("Correct!")
        score += 1
    else:
        print("Incorrect!")

print("\n========== QUIZ RESULT ==========")
print("Player:", name)
print("Score:", score, "/ 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Keep practising")

with open("scores.txt", "a") as file:
    file.write(f"{name},{score}/5\n")

print("Your score has been saved!")# Write your code here
````
