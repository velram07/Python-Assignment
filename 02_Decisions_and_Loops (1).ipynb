## 1. Indentation: Python's way of grouping code
Python uses **spaces at the start of a line** (indentation) to show which lines belong together - which lines are *inside* an `if` or a loop.
A line that ends with a colon `:` starts a block. Every line of that block must be indented by the same amount (normally **4 spaces**). Colab adds the spaces for you after you press Enter.

```
if marks >= 50:
    print("Pass")          # inside the if (indented)
    print("Well done")     # also inside the if
print("Result printed")    # outside the if (not indented) - always runs
```
Wrong or missing indentation gives an `IndentationError`.

## 2. The `if` statement
Remember the **diamond** in a flowchart from Week 2? `if` is that diamond in code. It asks a question (a condition). If the answer is `True`, the indented lines run; if it is `False`, they are skipped.

**Syntax:** `if condition:` followed by indented lines.

````python
marks = 72
if marks >= 50:              # the condition: is 72 >= 50? -> True
    print("Pass")            # runs only when the condition is True
print("Program finished")    # not indented, so it always runs
````

> **What happened?** The condition was True, so `Pass` was printed. Change `marks` to 40 and run again: `Pass` disappears, but `Program finished` is still printed because it is outside the `if`.

### `if ... else`: two paths
`else` gives a second path that runs when the condition is **False**. Exactly **one** of the two blocks always runs.

````python
attendance = 68
if attendance >= 75:                          # is 68 >= 75? -> False
    print("Eligible for the exam")            # skipped
else:                                         # runs because the condition was False
    print("Not eligible - attendance shortage")
````

> **What happened?** 68 is less than 75, so the `if` block was skipped and the `else` block ran. Note that `else` has no condition of its own and ends with a colon.

### `if ... elif ... else`: many paths
`elif` means **else if**. Use it when there are more than two possible results, like grades.
Python checks the conditions **from top to bottom** and runs only the **first** one that is True. The rest are skipped. `else` at the end catches everything that did not match.

````python
marks = 83

if marks >= 90:            # 83 >= 90? False -> check next
    grade = "O"
elif marks >= 80:          # 83 >= 80? True  -> this block runs, all others are skipped
    grade = "A+"
elif marks >= 70:
    grade = "A"
elif marks >= 60:
    grade = "B+"
elif marks >= 50:
    grade = "B"
else:                      # none of the above were True
    grade = "RA (re-appear)"

print("Grade:", grade)
````

> **What happened?** 83 is also >= 70, >= 60 and >= 50, but those lines were never checked, because Python stops at the **first** True condition. That is why the order matters: always check from the highest to the lowest.
> 
> **Try it:** change `marks` to 95, 61 and 40 and run the cell each time.

### Combining conditions
You can join conditions with `and` / `or` (from Notebook 1) inside an `if`.

````python
age = 19
has_licence = True

if age >= 18 and has_licence:      # BOTH must be True
    print("You can ride a two-wheeler to college")
else:
    print("Please take the college bus")
````

> **What happened?** Both conditions were True, so the first message was printed. Set `has_licence = False` and run again - now the `else` block runs.

### Nested `if`: an `if` inside another `if`
Sometimes a second question only makes sense after the first one is answered. Then we put one `if` **inside** another. Look at the indentation: the inner `if` is indented one more level.

````python
student_type = "hostel"
year = 1

if student_type == "hostel":           # first question
    if year == 1:                      # second question - only asked for hostel students
        print("Hostel fee for first years: Rs. 85,000")
    else:
        print("Hostel fee for seniors: Rs. 80,000")
else:                                  # day scholars
    print("Day scholar: college bus fee Rs. 30,000")
````

> **What happened?** Python first checked `student_type`. Because it was "hostel", it then checked `year` and printed the first-year fee. *(Fee amounts are sample values for practice.)*

### One-line `if` (conditional expression)
For a simple choice between two values, Python allows a short form:

`value_if_true if condition else value_if_false`

````python
n = 7
result = "even" if n % 2 == 0 else "odd"   # choose one of two words in one line
print(n, "is", result)
````

> **What happened?** `7 % 2` is 1, so the condition is False and `result` gets "odd". This is the same as a 4-line `if/else`, written in one line.

### Checking text with `in`
`in` asks whether one piece of text is **inside** another. `.lower()` changes the text to small letters first, so that `WON`, `Won` and `won` are all found.

````python
message = "Congratulations! You WON a free iPhone. Click the link"
if "won" in message.lower():                  # look for "won" in the lower-case message
    print("Warning: this message may be spam")
````

> **What happened?** The message contains `WON`. After `.lower()` it becomes `won`, so the condition is True and the warning appears. Without `.lower()` the check would fail, because Python treats capital and small letters as different.

## 4. The `while` loop
A loop **repeats** code. A `while` loop keeps repeating **as long as** its condition is True. Each repetition is called an **iteration**.

**Important:** something inside the loop must change, so that the condition eventually becomes False. Otherwise the loop never stops (an *infinite loop*).

````python
count = 5                       # start value
while count > 0:                # repeat while count is more than 0
    print("Launch in", count)
    count -= 1                  # reduce count by 1 each time - this makes the loop end
print("Lift off!")              # runs once, after the loop finishes
````

> **What happened?** The loop ran 5 times with count = 5, 4, 3, 2, 1. When count became 0, the condition `count > 0` was False, so the loop stopped and `Lift off!` was printed.
> 
> If you ever start an infinite loop in Colab, click the **stop** button next to the cell.

### `while` with user input: a PIN with 3 attempts
A `while` loop is perfect when we do not know in advance how many times to repeat - here, until the user types the correct PIN or runs out of attempts. `break` (explained in section 6) stops the loop immediately.

````python
correct_pin = "2468"
attempts = 0                         # how many times the user has tried

while attempts < 3:                  # allow at most 3 tries
    pin = input("Enter your UPI PIN: ")
    attempts += 1                    # count this attempt
    if pin == correct_pin:
        print("Payment successful")
        break                        # correct PIN - leave the loop now
    else:
        print("Wrong PIN. Attempts left:", 3 - attempts)

if attempts == 3 and pin != correct_pin:     # used all 3 tries and still wrong
    print("Account locked for 24 hours")
````

> **What happened?** Try it twice: once type `2468` (the loop ends at once with `break`), and once type three wrong PINs (the loop ends because `attempts < 3` becomes False, and the account is locked).

## 5. The `for` loop and `range()`
A `for` loop repeats once for **each item** in a sequence. It is the most common loop in Python.

`range()` creates a sequence of numbers:

| Code | Numbers produced |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 (starts at 0 and stops **before** 5) |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 (the third number is the step) |
| `range(10, 0, -1)` | 10, 9, ..., 1 (counting down) |

In `for i in range(5):`, the variable `i` takes each value in turn.

````python
for i in range(5):          # i becomes 0, then 1, 2, 3, 4
    print("Hello", i)       # this indented line runs once for each value of i
````

> **What happened?** The loop ran **5 times** and `i` went from 0 to 4 - not 1 to 5. `range(5)` starts at 0 and stops before 5.

Now a useful example: a **multiplication table**. We use `range(1, 11)` to get the numbers 1 to 10.

````python
n = 7
for i in range(1, 11):               # i = 1, 2, ..., 10
    print(f"{n} x {i:2} = {n * i}")  # {i:2} reserves 2 spaces so the columns line up
````

> **What happened?** One `print()` line inside the loop produced 10 lines of output. That is the power of loops: write once, repeat many times.

The next cell shows `range()` with a **step**: counting in twos, and counting backwards. `end=" "` keeps the numbers on one line.

````python
for even in range(2, 11, 2):      # start 2, stop before 11, step 2
    print(even, end=" ")
print()                           # an empty print() just moves to a new line

for x in range(10, 0, -1):        # start 10, stop before 0, step -1 (count down)
    print(x, end=" ")
````

> **What happened?** The first loop printed the even numbers 2 to 10. The second counted down from 10 to 1 - it stopped **before** 0.

### Looping over a string
A `for` loop can go through **each character** of a string, one at a time.

````python
word = "PYTHON"
for letter in word:     # letter becomes "P", then "Y", then "T", ...
    print(letter)
````

> **What happened?** The loop ran 6 times, once per letter. In Notebook 3 you will loop over lists and dictionaries in exactly the same way.

### `for` or `while`?
- Use **`for`** when you know how many times to repeat (10 times, every letter, every student in a list).
- Use **`while`** when you repeat **until something happens** (correct PIN entered, user types *quit*).

## 6. `break` and `continue`
Two keywords let you control a loop from inside:
- `break` - **stop** the whole loop immediately and jump to the code after it.
- `continue` - **skip** the rest of this round and go straight to the next round.

First, `break`: find the first number after 20 that is divisible by 7.

````python
for n in range(20, 40):        # check 20, 21, 22, ...
    if n % 7 == 0:             # divisible by 7?
        print("First multiple of 7 after 20 is", n)
        break                  # found it - no need to check the remaining numbers
````

> **What happened?** The loop checked 20 and 21... and found 21. `break` stopped the loop at once, so 28 and 35 were never printed.

Now `continue`: print study hours, but skip the days when the student did not study (marked 0).

````python
hours = [6, 0, 7, 5, 0, 8]     # a list of study hours for 6 days
for h in hours:
    if h == 0:
        continue               # skip the print below and move to the next day
    print("Studied", h, "hours")
````

> **What happened?** The two zero days were skipped, but the loop **kept going** - that is the difference from `break`.
> 
> `pass` is a third keyword. It means *do nothing* and is used as a placeholder when Python needs a line but you have not written the code yet.

## 7. Common loop patterns
These three patterns appear in almost every program. Learn them well!
1. **Accumulator** - start a total at 0 and add to it in each round.
2. **Counter** - start a count at 0 and add 1 when a condition is True.
3. **Find the largest** - remember the best value seen so far and replace it when you find a bigger one.

*(A list `[ ... ]` holds many values. You will learn lists in detail in Notebook 3.)*

````python
marks = [78, 45, 92, 66, 38, 85]

# Pattern 1: accumulator (running total)
total = 0                     # start with nothing
for m in marks:
    total += m                # add each mark to the total
print("Total:", total)

# Pattern 2: counter
passed = 0                    # start counting from 0
for m in marks:
    if m >= 50:
        passed += 1           # count only the marks that pass
print("Students passed:", passed)

# Pattern 3: find the largest
highest = marks[0]            # assume the first mark is the highest
for m in marks:
    if m > highest:           # found a bigger one?
        highest = m           # remember it
print("Highest:", highest)
````

> **What happened?** Total = 404, 4 students passed (78, 92, 66, 85), and the highest mark is 92. Python also has the built-in functions `sum()` and `max()` for this, but you must understand these patterns, because many problems need a version that no built-in function covers.

### AI connection: calculating accuracy with a loop
In Week 3 you counted correct predictions by hand. Here a loop compares the **actual** labels with the model's **predicted** labels, one by one, and counts the matches. `len()` gives the number of items in a list.

````python
actual    = ["dry", "wet", "dry", "wet", "dry", "wet", "dry", "wet", "dry", "wet"]
predicted = ["dry", "wet", "wet", "wet", "dry", "wet", "dry", "dry", "dry", "wet"]

correct = 0
for i in range(len(actual)):           # i = 0, 1, ..., 9 (one index per test image)
    if actual[i] == predicted[i]:      # compare the two labels at position i
        correct += 1                   # counter pattern

print(f"Correct: {correct} / {len(actual)}")
print(f"Accuracy: {correct / len(actual) * 100:.0f}%")
````

> **What happened?** Images 3 and 8 were predicted wrongly, so 8 out of 10 are correct: accuracy = 80%. Real AI libraries calculate accuracy in exactly this way, just on thousands of examples.

## 8. Nested loops
A **nested loop** is a loop inside another loop. The inner loop runs **completely** for **each** round of the outer loop. It is useful for patterns, tables and grids.

````python
rows = 5
for i in range(1, rows + 1):      # outer loop: one round per row (i = 1 to 5)
    for j in range(i):            # inner loop: print i stars in this row
        print("*", end=" ")
    print()                       # after each row, move to a new line
````

> **What happened?** In row 1 the inner loop ran once, in row 2 twice, and so on, making a triangle. The empty `print()` belongs to the **outer** loop (look at its indentation), so it runs once per row.

Python can also **repeat a string** with `*`: `"* " * 3` gives `"* * * "`. This makes the same pattern with only one loop.

````python
for i in range(1, 6):
    print("* " * i)       # repeat "* " i times
````

> **What happened?** Same triangle, shorter code. There is often more than one correct way to solve a problem in Python.

## 9. Common mistakes
| Mistake | Example | Fix |
|---|---|---|
| `=` instead of `==` in a condition | `if marks = 50:` | `if marks == 50:` |
| Forgetting the colon | `if marks > 50` | `if marks > 50:` |
| Wrong indentation | body of `if` not indented | indent 4 spaces |
| Infinite `while` loop | forgetting `count -= 1` | change the variable inside the loop; press the **stop** button in Colab |
| Off-by-one in `range` | `range(1, 10)` to print 1 to 10 | `range(1, 11)` - the stop value is not included |

- `range(start, stop, step)` stops **before** `stop`.
- `break` exits a loop; `continue` skips to the next round.

## Mini project: Number guessing game
**Mini project.** The computer picks a secret number from 1 to 50 using `random.randint(1, 50)`. The player gets **at most 7 guesses**. After each guess print *Too high*, *Too low* or *Correct!*. At the end, print how many guesses were used, or reveal the number if the player lost.

Start with:
```python

```

<details><summary><b>Hint</b> (click to open)</summary>

Use a `while guesses < 7:` loop. Inside: read a guess, add 1 to `guesses`, and use `if/elif/else` to compare with `secret`. Use `break` when the guess is correct and a `won` flag to decide the final message.

````python
import random                     # random is a ready-made module (Notebook 4)
secret = random.randint(1, 50)    # a random whole number from 1 to 50

import random

secret = random.randint(1, 50)
guesses = 0
won = False

print("Welcome to the Number Guessing Game!")
print("Guess a number between 1 and 50.")
print("You have 7 attempts.")

for i in range(7):
    guess = int(input("Enter your guess: "))
    guesses += 1

    if guess == secret:
        print("Correct!")
        won = True
        break
    elif guess > secret:
        print("Too high")
    else:
        print("Too low")

print("Guesses used:", guesses)

if not won:
    print("You lost! The number was:", secret)
````
